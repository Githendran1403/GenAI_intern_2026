## Problem 1: Employee Performance Bonus Eligibility

**Description:** A company evaluates employee performance scores at the end of the year.

**Requirements:**
- Identify the highest performance score
- Handle ties if multiple employees have the same highest score
- Display all employees eligible for the top performance bonus

```python
# Problem 1: Employee Performance Bonus Eligibility

employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Find the highest score
highest_score = max(employees.values())

# Find all employees with highest score
top_performers = []
for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

# Display result
print(f"Top Performers Eligible for Bonus: {', '.join(top_performers)} (Score: {highest_score})")
```

## Problem 2: Search Query Keyword Analysis

**Description:** An e-commerce website stores customer search queries.

**Requirements:**
- Convert the input to lowercase
- Ignore common punctuation
- Count the frequency of each keyword
- Display only keywords searched more than once

```python
# Problem 2: Search Query Keyword Analysis

query = "Buy mobile phone buy phone online"

# Convert to lowercase
query = query.lower()

# Split into words
words = query.split()

# Count frequency of each word
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Filter words that appear more than once
repeated_words = {}
for word, count in word_count.items():
    if count > 1:
        repeated_words[word] = count

# Display result
print(repeated_words)
```

## Problem 3: Sensor Data Validation

**Description:** A factory collects sensor readings every hour.

**Requirements:**
- Identify readings that are even numbers (valid readings)
- Store them as (hour_index, reading_value) pairs
- Ignore odd readings (invalid readings)

```python
# Problem 3: Sensor Data Validation

sensor_readings = [3, 4, 7, 8, 10, 12, 5]

# Find valid readings (even numbers)
valid_readings = []
for index, reading in enumerate(sensor_readings):
    if reading % 2 == 0:  # Check if even
        valid_readings.append((index, reading))

# Display result
print(f"Valid Sensor Readings (Hour, Value): {valid_readings}")
```

## Problem 4: Email Domain Usage Analysis

**Description:** A company wants to analyze which email providers its users are using.

**Requirements:**
- Count how many users belong to each email domain
- Calculate the percentage usage of each domain

```python
# Problem 4: Email Domain Usage Analysis

emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

# Count domains
domain_count = {}
for email in emails:
    # Extract domain (part after @)
    domain = email.split('@')[1]
    
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

# Calculate and display percentages
total_emails = len(emails)
for domain, count in domain_count.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {percentage:.0f}%")
```

## Problem 5: Sales Spike Detection

**Description:** A retail company tracks daily sales. Sudden spikes in sales may indicate promotions or unusual activity.

**Requirements:**
- Calculate the average daily sales
- Detect days where sales are more than 30% above average
- Display the day number and sale value

```python
# Problem 5: Sales Spike Detection

sales = [1200, 1500, 900, 2200, 1400, 3000]

# Calculate average sales
total_sales = 0
for sale in sales:
    total_sales += sale
average_sales = total_sales / len(sales)

# Calculate threshold (30% above average)
threshold = average_sales * 1.30

# Detect spikes
for day, sale in enumerate(sales, start=1):
    if sale > threshold:
        print(f"Day {day}: {sale}")
```

## Problem 6: Duplicate User ID Detection

**Description:** A system stores user IDs during registration. Duplicate IDs can cause data integrity issues.

**Requirements:**
- Identify duplicate user IDs
- Display how many times each duplicate appears

```python
# Problem 6: Duplicate User ID Detection

user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Count occurrences of each user ID
id_count = {}
for user_id in user_ids:
    if user_id in id_count:
        id_count[user_id] += 1
    else:
        id_count[user_id] = 1

# Display only duplicates (count > 1)
for user_id, count in id_count.items():
    if count > 1:
        print(f"{user_id} → {count} times")
```
