## Problem 1: Smart Parking Lot Management System

**Problem Statement:** Design a function to manage a smart parking lot.

**Requirements:**
- Accept vehicle entry and exit logs
- Calculate total parked vehicles
- Identify peak parking usage
- Alert if parking exceeds capacity

**Real-Time Use:** Mall parking systems, Smart city infrastructure

```python
# Problem 1: Smart Parking Lot Management System

def manage_parking(capacity, vehicle_logs):
    """
    Manage parking lot and track vehicle entries/exits
    """
    currently_parked = 0
    
    # Process vehicle logs
    for log in vehicle_logs:
        if log == "IN":
            currently_parked += 1
        elif log == "OUT":
            currently_parked -= 1
    
    # Determine parking status
    if currently_parked >= capacity:
        status = "Full"
    elif currently_parked >= capacity * 0.8:
        status = "Almost Full"
    else:
        status = "Available"
    
    return currently_parked, status

# Test the function
capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

parked, status = manage_parking(capacity, vehicle_logs)

print("=" * 50)
print("SMART PARKING LOT MANAGEMENT")
print("=" * 50)
print(f"Parking Capacity: {capacity}")
print(f"Currently Parked Vehicles: {parked}")
print(f"Parking Status: {status}")
print("=" * 50)
```

## Problem 2: Online Food Delivery Time Estimator

**Problem Statement:** Create a function that estimates delivery time based on distance, weather, and traffic.

**Requirements:**
- Distance (km)
- Weather condition
- Traffic level
- Apply delays dynamically and display final ETA

**Real-Time Use:** Food delivery apps, Logistics platforms

```python
# Problem 2: Online Food Delivery Time Estimator

def estimate_delivery_time(distance, traffic_level, weather):
    """
    Estimate delivery time based on multiple factors
    """
    # Base time: 5 minutes per km
    base_time = distance * 5
    
    # Traffic delay
    if traffic_level == "High":
        traffic_delay = 15
    elif traffic_level == "Medium":
        traffic_delay = 8
    else:
        traffic_delay = 0
    
    # Weather delay
    if weather == "Rainy":
        weather_delay = 10
    elif weather == "Stormy":
        weather_delay = 20
    else:
        weather_delay = 0
    
    # Calculate total time
    total_time = base_time + traffic_delay + weather_delay
    
    return total_time

# Test the function
distance = 8
traffic = "High"
weather = "Rainy"

eta = estimate_delivery_time(distance, traffic, weather)

print("=" * 50)
print("FOOD DELIVERY TIME ESTIMATOR")
print("=" * 50)
print(f"Distance: {distance} km")
print(f"Traffic Level: {traffic}")
print(f"Weather: {weather}")
print(f"Estimated Delivery Time: {eta} minutes")
print("=" * 50)
```

## Problem 3: Movie Theatre Seat Occupancy Analyzer

**Problem Statement:** Build a function that analyzes seat booking data.

**Requirements:**
- Calculates occupancy percentage
- Determines if show is Housefull
- Suggests opening additional shows

**Real-Time Use:** Cinema ticketing systems, Event management software

```python
# Problem 3: Movie Theatre Seat Occupancy Analyzer

def analyze_seat_occupancy(total_seats, booked_seats):
    """
    Analyze theatre seat occupancy
    """
    # Count booked seats
    num_booked = len(booked_seats)
    
    # Calculate occupancy percentage
    occupancy = (num_booked / total_seats) * 100
    
    # Determine show status
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    elif occupancy >= 50:
        status = "Good Booking"
    else:
        status = "Available"
    
    return occupancy, status

# Test the function
total_seats = 200
booked_seats = [1] * 150  # 150 booked seats

occupancy, status = analyze_seat_occupancy(total_seats, booked_seats)

print("=" * 50)
print("MOVIE THEATRE SEAT OCCUPANCY ANALYZER")
print("=" * 50)
print(f"Total Seats: {total_seats}")
print(f"Booked Seats: {len(booked_seats)}")
print(f"Occupancy: {occupancy:.0f}%")
print(f"Show Status: {status}")
print("=" * 50)
```

## Problem 4: Cloud Server Load Classification System

**Problem Statement:** Create a function to classify server load based on CPU usage readings.

**Rules:**
- Average CPU < 50% → Normal
- 50%–80% → Warning
- > 80% → Critical

**Real-Time Use:** Cloud monitoring dashboards, DevOps alerting systems

```python
# Problem 4: Cloud Server Load Classification System

def classify_server_load(cpu_readings):
    """
    Classify server load based on CPU usage
    """
    # Calculate average CPU
    total = 0
    for reading in cpu_readings:
        total += reading
    
    average_cpu = total / len(cpu_readings)
    
    # Classify server status
    if average_cpu > 80:
        status = "Critical"
    elif average_cpu >= 50:
        status = "Warning"
    else:
        status = "Normal"
    
    return average_cpu, status

# Test the function
cpu_readings = [45, 60, 70, 85, 90]

avg_cpu, status = classify_server_load(cpu_readings)

print("=" * 50)
print("CLOUD SERVER LOAD CLASSIFICATION")
print("=" * 50)
print(f"CPU Readings (%): {cpu_readings}")
print(f"Average CPU Load: {avg_cpu:.0f}%")
print(f"Server Status: {status}")
print("=" * 50)
```

## Problem 5: Smart Classroom Resource Usage Monitor

**Problem Statement:** Design a function that tracks usage of classroom resources and identifies overuse patterns.

**Requirements:**
- Track projector, AC, lights usage
- Identify overused resources (> 8 hours)
- Trigger energy alert

**Real-Time Use:** Smart classrooms, Energy optimization systems

```python
# Problem 5: Smart Classroom Resource Usage Monitor

def monitor_resource_usage(resource_usage):
    """
    Monitor classroom resource usage and detect overuse
    """
    overused_resources = []
    threshold = 8  # hours
    
    # Check each resource
    for resource, hours in resource_usage.items():
        if hours > threshold:
            overused_resources.append(resource)
    
    # Determine energy alert
    if len(overused_resources) > 0:
        energy_alert = "Yes"
    else:
        energy_alert = "No"
    
    return overused_resources, energy_alert

# Test the function
resource_usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

overused, alert = monitor_resource_usage(resource_usage)

print("=" * 50)
print("SMART CLASSROOM RESOURCE MONITOR")
print("=" * 50)
print("Resource Usage (hours):")
for resource, hours in resource_usage.items():
    print(f"  {resource}: {hours} hours")
print("-" * 50)
if overused:
    print(f"Overused Resources: {', '.join(overused)}")
else:
    print("Overused Resources: None")
print(f"Energy Alert: {alert}")
print("=" * 50)
```

## Problem 6: Online Event Registration Capacity Controller

**Problem Statement:** Create a function that manages event registrations.

**Requirements:**
- Tracking registrations
- Preventing overbooking
- Triggering waitlist mode

**Real-Time Use:** Webinar platforms, Conference registration systems

```python
# Problem 6: Online Event Registration Capacity Controller

def manage_event_registration(capacity, registrations):
    """
    Manage event registrations and handle waitlist
    """
    # Calculate confirmed and waitlisted
    if registrations <= capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    
    return confirmed, waitlisted, status

# Test the function
event_capacity = 100
total_registrations = 105

confirmed, waitlisted, status = manage_event_registration(event_capacity, total_registrations)

print("=" * 50)
print("EVENT REGISTRATION CAPACITY CONTROLLER")
print("=" * 50)
print(f"Event Capacity: {event_capacity}")
print(f"Total Registrations: {total_registrations}")
print("-" * 50)
print(f"Confirmed Registrations: {confirmed}")
print(f"Waitlisted Users: {waitlisted}")
print(f"Registration Status: {status}")
print("=" * 50)
```
