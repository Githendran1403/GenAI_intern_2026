## Problem 1: Mobile Recharge Validation System

**Problem Statement:** Create a function to validate a mobile recharge request.

**Rules:**
- Recharge amount must be ≥ ₹50
- Valid plans: 199, 299, 399, 599
- User should be allowed to retry if invalid

**Real-Time Use:** Telecom recharge apps, Payment gateways

```python
# Problem 1: Mobile Recharge Validation System

def validate_recharge(amount):
    """
    Validate mobile recharge amount
    """
    valid_plans = [199, 299, 399, 599]
    
    if amount < 50:
        return False, "Recharge amount must be at least ₹50"
    elif amount in valid_plans:
        return True, f"Recharge of ₹{amount} successful!"
    else:
        return False, f"Invalid plan. Valid plans are: {valid_plans}"

# Test with retry loop
print("=" * 50)
print("MOBILE RECHARGE SYSTEM")
print("=" * 50)

while True:
    amount = int(input("Enter recharge amount (or 0 to exit): ₹"))
    
    if amount == 0:
        print("Thank you for using our service!")
        break
    
    is_valid, message = validate_recharge(amount)
    print(message)
    
    if is_valid:
        break
    else:
        print("Please try again.\n")
```

## Problem 2: Inventory Reorder Alert System

**Problem Statement:** Given a dictionary of products and stock quantities, print alerts for low stock.

**Rules:**
- Print "Reorder Alert" if stock < 15
- Print "Stock OK" otherwise

**Real-Time Use:** Warehouse systems, Retail management

```python
# Problem 2: Inventory Reorder Alert System

def check_inventory(inventory):
    """
    Check inventory and alert for low stock
    """
    print("=" * 50)
    print("INVENTORY REORDER ALERT SYSTEM")
    print("=" * 50)
    
    for product, stock in inventory.items():
        if stock < 15:
            print(f"{product}: {stock} units - ⚠️ REORDER ALERT")
        else:
            print(f"{product}: {stock} units - ✓ Stock OK")

# Test the function
inventory = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 8,
    "Monitor": 20,
    "Headphones": 12
}

check_inventory(inventory)
```

## Problem 3: Student Result Processing System

**Problem Statement:** Create a function that processes student marks.

**Requirements:**
- Accepts a list of student marks
- Calculates average marks
- Prints "Pass" if average ≥ 50 else "Fail"

**Real-Time Use:** Online exam systems, Evaluation portals

```python
# Problem 3: Student Result Processing System

def process_result(marks):
    """
    Process student marks and determine result
    """
    # Calculate total
    total = 0
    for mark in marks:
        total += mark
    
    # Calculate average
    average = total / len(marks)
    
    # Determine result
    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"
    
    return average, result

# Test the function
student_marks = [45, 67, 52, 78, 60]

average, result = process_result(student_marks)

print("=" * 40)
print("STUDENT RESULT PROCESSING")
print("=" * 40)
print(f"Marks: {student_marks}")
print(f"Average: {average:.2f}")
print(f"Result: {result}")
print("=" * 40)
```

## Problem 4: Cab Fare Estimator with Retry Option

**Problem Statement:** Create a cab fare calculator.

**Rules:**
- Base fare = ₹50
- ₹12 per km
- Peak hour → 25% extra
- Allow retry using while loop

**Real-Time Use:** Ride-hailing apps, Transport systems

```python
# Problem 4: Cab Fare Estimator with Retry Option

def calculate_fare(distance, is_peak_hour):
    """
    Calculate cab fare based on distance and peak hour
    """
    base_fare = 50
    per_km_rate = 12
    
    # Calculate basic fare
    fare = base_fare + (distance * per_km_rate)
    
    # Add peak hour charges
    if is_peak_hour:
        peak_charge = fare * 0.25
        fare = fare + peak_charge
        return fare, peak_charge
    else:
        return fare, 0

# Test with retry loop
print("=" * 50)
print("CAB FARE ESTIMATOR")
print("=" * 50)

while True:
    distance = float(input("Enter distance in km (or 0 to exit): "))
    
    if distance == 0:
        print("Thank you!")
        break
    
    peak = input("Is it peak hour? (yes/no): ").lower()
    is_peak = True if peak == "yes" else False
    
    fare, peak_charge = calculate_fare(distance, is_peak)
    
    print(f"\nDistance: {distance} km")
    if is_peak:
        print(f"Peak Hour Charge: ₹{peak_charge:.2f}")
    print(f"Total Fare: ₹{fare:.2f}")
    print("\n" + "=" * 50)
    
    retry = input("Calculate another fare? (yes/no): ").lower()
    if retry != "yes":
        print("Thank you!")
        break
```

## Problem 5: Employee Attendance Eligibility Checker

**Problem Statement:** Create a function to check attendance eligibility.

**Requirements:**
- Accepts attendance list ("P", "A")
- Calculates attendance percentage
- Returns "Eligible" if ≥ 75% else "Not Eligible"

**Real-Time Use:** HR payroll systems, Internship tracking

```python
# Problem 5: Employee Attendance Eligibility Checker

def check_attendance_eligibility(attendance):
    """
    Check if employee is eligible based on attendance
    """
    # Count present days
    present_count = 0
    for day in attendance:
        if day == "P":
            present_count += 1
    
    # Calculate percentage
    total_days = len(attendance)
    percentage = (present_count / total_days) * 100
    
    # Check eligibility
    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"
    
    return percentage, status

# Test the function
attendance_record = ["P", "P", "A", "P", "P", "P", "A", "P", "P", "P"]

percentage, status = check_attendance_eligibility(attendance_record)

print("=" * 40)
print("ATTENDANCE ELIGIBILITY CHECKER")
print("=" * 40)
print(f"Total Days: {len(attendance_record)}")
print(f"Attendance: {percentage:.2f}%")
print(f"Status: {status}")
print("=" * 40)
```

## Problem 6: Password Strength Checker

**Problem Statement:** Create a function to check password strength.

**Rules:**
- Minimum length = 8
- Must contain at least one digit
- Must contain at least one special character (@#$)

**Real-Time Use:** Login systems, Security validation

```python
# Problem 6: Password Strength Checker

def check_password_strength(password):
    """
    Check if password meets strength requirements
    """
    # Check minimum length
    if len(password) < 8:
        return False, "Password must be at least 8 characters long"
    
    # Check for digit
    has_digit = False
    for char in password:
        if char.isdigit():
            has_digit = True
            break
    
    if not has_digit:
        return False, "Password must contain at least one digit"
    
    # Check for special character
    special_chars = "@#$"
    has_special = False
    for char in password:
        if char in special_chars:
            has_special = True
            break
    
    if not has_special:
        return False, "Password must contain at least one special character (@#$)"
    
    return True, "Password is strong!"

# Test the function
test_passwords = [
    "weak",
    "password123",
    "Pass@123",
    "Strong#99"
]

print("=" * 50)
print("PASSWORD STRENGTH CHECKER")
print("=" * 50)

for pwd in test_passwords:
    is_strong, message = check_password_strength(pwd)
    status = "✓ STRONG" if is_strong else "✗ WEAK"
    print(f"Password: '{pwd}'")
    print(f"Status: {status}")
    print(f"Message: {message}")
    print("-" * 50)
```
