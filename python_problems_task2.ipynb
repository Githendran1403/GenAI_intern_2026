## Problem 1: Count Unique Words

**Problem:** Given a sentence, identify all unique words and display the count and the unique words.

**Real-Time Applications:** Text analytics, NLP preprocessing, Search engine indexing

```python
# Problem 1: Count Unique Words

sentence = "python is easy and python is powerful"

# Split sentence into words
words = sentence.split()

# Convert to set to get unique words
unique_words = set(words)

# Display results
print(f"Unique words count: {len(unique_words)}")
print(f"Unique words: {unique_words}")
```

## Problem 2: Highest Salary from Employee Data

**Problem:** Store employee names and their salaries, and determine which employee has the highest salary.

**Real-Time Applications:** HR management systems, Payroll analysis

```python
# Problem 2: Highest Salary from Employee Data

employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# Find employee with highest salary
highest_employee = max(employees, key=employees.get)
highest_salary = employees[highest_employee]

# Display result
print(f"Highest Salary: {highest_employee} - {highest_salary}")
```

## Problem 3: Find Maximum and Minimum Values

**Problem:** Find the maximum and minimum values from a list of numbers.

**Real-Time Applications:** Sales data analysis, Temperature monitoring, Performance metrics

```python
# Problem 3: Find Maximum and Minimum Values

numbers = [45, 22, 89, 10, 66]

# Initialize max and min with first element
maximum = numbers[0]
minimum = numbers[0]

# Loop through list to find max and min
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

# Display results
print(f"List: {numbers}")
print(f"Max: {maximum}")
print(f"Min: {minimum}")
```

## Problem 4: Count Products Above a Price Threshold

**Problem:** Count how many products have a price greater than 1000.

**Real-Time Applications:** E-commerce product filtering, Budget analysis

```python
# Problem 4: Count Products Above a Price Threshold

prices = [450, 1200, 899, 1500, 300]

# Count products above 1000
count = 0
for price in prices:
    if price > 1000:
        count += 1

# Display result
print(f"Products above 1000: {count}")
```

## Problem 5: Calculate Attendance Percentage

**Problem:** Calculate the attendance percentage from a list containing "P" (Present) and "A" (Absent).

**Real-Time Applications:** Internship tracking, HR attendance management

```python
# Problem 5: Calculate Attendance Percentage

attendance = ["P", "P", "A", "P", "P"]

# Count present days
present_count = 0
for day in attendance:
    if day == "P":
        present_count += 1

# Calculate percentage
total_days = len(attendance)
percentage = (present_count / total_days) * 100

# Display result
print(f"Attendance Percentage: {percentage}")
```

## Problem 6: Remove Duplicate Phone Numbers

**Problem:** Remove duplicate phone numbers and display only unique phone numbers.

**Real-Time Applications:** Contact management, CRM systems, Marketing databases

```python
# Problem 6: Remove Duplicate Phone Numbers

phone_numbers = [9876543210, 9123456789, 9876543210]

# Convert to set to remove duplicates
unique_numbers = set(phone_numbers)

# Display result
print(f"Unique phone numbers: {unique_numbers}")
```

## Problem 7: Count Character Frequency

**Problem:** Count how many times each character appears in a given string.

**Real-Time Applications:** Text analysis, Password strength validation

```python
# Problem 7: Count Character Frequency

text = "python"

# Create empty dictionary
char_count = {}

# Count each character
for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1

# Display result
print(char_count)
```

## Problem 9: Check if a Key Exists in a Dictionary

**Problem:** Check whether a specific key exists in a dictionary.

**Real-Time Applications:** User validation systems, Configuration verification

```python
# Problem 9: Check if a Key Exists in a Dictionary

employee_data = {
    "name": "Ravi",
    "age": 25,
    "salary": 50000
}

# Check if key exists
key_to_check = "name"

if key_to_check in employee_data:
    print("Employee exists")
else:
    print("Employee does not exist")
```

## Problem 10: Calculate Average Marks

**Problem:** Calculate the average marks of a student from a list of marks.

**Real-Time Applications:** Student performance evaluation, Result processing systems

```python
# Problem 10: Calculate Average Marks

marks = [80, 75, 90, 65]

# Calculate sum
total = 0
for mark in marks:
    total += mark

# Calculate average
average = total / len(marks)

# Display result
print(f"Average Marks: {average}")
```
