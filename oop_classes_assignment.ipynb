## Problem 1: Smart Light Controller

**Problem Statement:** Create a class to control a smart light that can be turned ON or OFF and display its status.

**Real-Time Use:** Smart home automation systems

```python
# Problem 1: Smart Light Controller

class SmartLight:
    def __init__(self, name):
        """
        Initialize smart light with name
        """
        self.name = name
        self.status = "OFF"  # Default status
    
    def turn_on(self):
        """
        Turn the light ON
        """
        self.status = "ON"
    
    def turn_off(self):
        """
        Turn the light OFF
        """
        self.status = "OFF"
    
    def display_status(self):
        """
        Display current light status
        """
        print(f"{self.name} is {self.status}")

# Test the class
print("=" * 50)
print("SMART LIGHT CONTROLLER")
print("=" * 50)

light = SmartLight("Bedroom Light")
light.turn_on()
light.display_status()

print()
light.turn_off()
light.display_status()
```

## Problem 2: Employee ID Card System

**Problem Statement:** Create a class that stores employee name, ID, and department and displays ID card details.

**Real-Time Use:** Corporate HR systems

```python
# Problem 2: Employee ID Card System

class Employee:
    def __init__(self, name, emp_id, department):
        """
        Initialize employee with name, ID, and department
        """
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    def display_id_card(self):
        """
        Display employee ID card details
        """
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")

# Test the class
print("=" * 50)
print("EMPLOYEE ID CARD SYSTEM")
print("=" * 50)

employee = Employee("Rahul", "EMP102", "IT")
employee.display_id_card()
```

## Problem 3: Mobile Contact Record

**Problem Statement:** Create a class to store contact name and phone number and display contact information.

**Real-Time Use:** Mobile contact applications

```python
# Problem 3: Mobile Contact Record

class Contact:
    def __init__(self, name, phone_number):
        """
        Initialize contact with name and phone number
        """
        self.name = name
        self.phone_number = phone_number
    
    def save_contact(self):
        """
        Save and display contact information
        """
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone_number}")

# Test the class
print("=" * 50)
print("MOBILE CONTACT RECORD")
print("=" * 50)

contact = Contact("Anita", "9876543210")
contact.save_contact()
```

## Problem 4: Product Price Tag Generator

**Problem Statement:** Create a class that stores product name and price and prints a formatted price tag.

**Real-Time Use:** Retail billing systems

```python
# Problem 4: Product Price Tag Generator

class Product:
    def __init__(self, name, price):
        """
        Initialize product with name and price
        """
        self.name = name
        self.price = price
    
    def generate_price_tag(self):
        """
        Generate and display formatted price tag
        """
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")

# Test the class
print("=" * 50)
print("PRODUCT PRICE TAG GENERATOR")
print("=" * 50)

product = Product("Headphones", 2499)
product.generate_price_tag()
```

## Problem 5: Movie Rating Display System

**Problem Statement:** Create a class that stores movie name and rating and displays movie details.

**Real-Time Use:** Streaming platforms

```python
# Problem 5: Movie Rating Display System

class Movie:
    def __init__(self, name, rating):
        """
        Initialize movie with name and rating
        """
        self.name = name
        self.rating = rating
    
    def display_details(self):
        """
        Display movie details with rating
        """
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating} / 5")

# Test the class
print("=" * 50)
print("MOVIE RATING DISPLAY SYSTEM")
print("=" * 50)

movie = Movie("Inception", 4.8)
movie.display_details()
```

## Problem 6: Delivery Address Manager

**Problem Statement:** Create a class that stores customer name and delivery address and prints delivery details.

**Real-Time Use:** Courier and logistics platforms

```python
# Problem 6: Delivery Address Manager

class Delivery:
    def __init__(self, customer_name, address):
        """
        Initialize delivery with customer name and address
        """
        self.customer_name = customer_name
        self.address = address
    
    def display_delivery_details(self):
        """
        Display delivery details
        """
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")

# Test the class
print("=" * 50)
print("DELIVERY ADDRESS MANAGER")
print("=" * 50)

delivery = Delivery("Suman", "Hyderabad")
delivery.display_delivery_details()
```
